# 导出 BI 汇总数据

本 Notebook 只负责导出前两部分已经生成的聚合结果，不重复执行分析和绘图。

```python
from pathlib import Path
import duckdb


PROJECT_DIR = Path.cwd()
DB_PATH = PROJECT_DIR / "data" / "criteo_analysis.duckdb"
BI_DIR = PROJECT_DIR / "reports" / "tableau"

if not DB_PATH.exists():
    raise FileNotFoundError(
        f"未找到数据库：{DB_PATH}\n"
        "请先依次运行 01_business_analysis.ipynb 和 02_attribution_decision.ipynb。"
    )

BI_DIR.mkdir(parents=True, exist_ok=True)
con = duckdb.connect(str(DB_PATH))

bi_tables = [
    "funnel_kpi",
    "daily_trend",
    "campaign_stats",
    "campaign_actions",
    "campaign_action_summary",
    "campaign_gap",
    "position_credit",
    "position_bid_multipliers",
    "bid_scope",
    "ipw_scenarios",
    "ipw_rank_stability",
    "position_reweight_sensitivity",
    "frequency_buckets",
    "attribution_window",
    "official_window_rule",
    "path_coverage",
    "path_length_distribution",
]

for table_name in bi_tables:
    output_path = (BI_DIR / f"{table_name}.csv").as_posix()

    con.execute(f"""
        COPY {table_name}
        TO '{output_path}'
        (HEADER, DELIMITER ',')
    """)

print(f"已导出 {len(bi_tables)} 张 BI 汇总数据表：{BI_DIR}")
```

```python
con.close()
print("数据库连接已关闭。")
```
